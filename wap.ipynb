```python
#wap to count vowels in string using functions in py
def count_vowels(text):
    vowels = "aeiouAEIOU"
    count = 0
    for ch in text:
        if ch in vowels:
            count += 1
    return count

# Example
s = "Hello World"
print("Vowel count:", count_vowels(s))
```

```python
#calc frequency of each character using functions in python
def char_frequency(text):
    freq = {}
    for ch in text:
        if ch in freq:
            freq[ch] += 1
        else:
            freq[ch] = 1
    return freq

# Example
s = "hello world"
print(char_frequency(s))
```

```python
# Linear Search
def linear_search(arr, x):
    for i in range(len(arr)):
        if arr[i] == x:
            return i
    return -1

# Binary Search (sorted list)
def binary_search(arr, x):
    low, high = 0, len(arr)-1
    while low <= high:
        mid = (low + high)//2
        if arr[mid] == x:
            return mid
        elif arr[mid] < x:
            low = mid + 1
        else:
            high = mid - 1
    return -1

# Example
nums = [10, 20, 30, 40, 50]
print("Linear:", linear_search(nums, 30))   # 2
print("Binary:", binary_search(nums, 30))   # 2
```

```python

# find the largest value in the list in py
def find_max(lst):
    max_val = lst[0]   # assume first element is max
    for num in lst:
        if num > max_val:
            max_val = num
    return max_val

# Example
nums = [12, 45, 7, 89, 23]
print("Largest value:", find_max(nums))
```

```python
#wap to remove duplicate elements in list
def remove_duplicates(lst):
    return list(set(lst))

# Example
nums = [1, 2, 3, 2, 4, 1, 5]
print("Without duplicates:", remove_duplicates(nums))
```

```python
#a program to find second largest no in the list 
def second_largest(lst):
    lst.sort()              # sort the list
    return lst[-2]          # second last element is 2nd largest

# Example
nums = [12, 45, 7, 89, 23]
print("Second largest:", second_largest(nums))
```

```python
def are_anagrams(s1, s2):
    return sorted(s1) == sorted(s2)

# Example
str1 = "listen"
str2 = "silent"
print("Anagram:", are_anagrams(str1, str2))
```

```python
#count words in the sentenaces
def count_words(sentence):
    words = sentence.split()   # split by spaces
    return len(words)

# Example
s = "Python is very easy to learn"
print("Word count:", count_words(s))
```
