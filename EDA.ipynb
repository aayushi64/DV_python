# 📊 Exploratory Data Analysis (EDA)

## 🔹 What is EDA?
Exploratory Data Analysis (EDA) is the process of analyzing datasets to:
- Understand structure (rows, columns, data types)
- Summarize statistics (mean, median, min, max, std)
- Detect missing values, duplicates, and outliers
- Identify distributions and correlations
- Prepare data for modeling

---

## 🔹 Why EDA is Used
- ✅ Gain insights before modeling
- ✅ Ensure data quality
- ✅ Spot anomalies
- ✅ Guide feature selection
- ✅ Support better decisions

---

## 🔹 Common EDA Techniques
- **Descriptive statistics** → `df.describe()`, `df.info()`
- **Data visualization** → histograms, scatter plots, box plots
- **Correlation analysis** → heatmaps, pairplots
- **Missing values check** → `df.isnull().sum()`
- **Value counts** → `df['column'].value_counts()`

---

import pandas as pd

df.loc[(df['Age'] < 0) | (df['Age'] > 100), 'Age'] = df['Age'].median()

# Example: Marks should be between 0 and 100
df.loc[(df['Marks'] < 0) | (df['Marks'] > 100), 'Marks'] = df['Marks'].mean()

# Pandas groupby()

## 🔹 Definition
- `groupby()` ek **split-apply-combine** operation hai.
- Matlab: Data ko **groups mein split** karo → har group pe ek function apply karo → phir results ko combine karke ek naya DataFrame/Series banao.

---

## 🔹 Steps
1. **Split** → Data ko ek ya multiple columns ke basis pe groups mein todta hai.  
2. **Apply** → Har group pe aggregation function lagata hai (mean, sum, count, etc.).  
3. **Combine** → Results ko ek structured output mein merge karta hai.

---

## 🔹 Common Aggregations
- `.mean()` → average per group  
- `.sum()` → total per group  
- `.count()` → number of rows per group  
- `.max()` / `.min()` → highest/lowest per group  

---

## 🔹 Example

```python
import pandas as pd

data = {
    'Department': ['CSE', 'CSE', 'ECE', 'ECE', 'ME'],
    'Student': ['A', 'B', 'C', 'D', 'E'],
    'Marks': [85, 90, 78, 88, 92]
}

df = pd.DataFrame(data)

# Group by Department and calculate mean marks
result = df.groupby('Department')['Marks'].mean()
print(result)
