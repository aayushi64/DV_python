# 📊 Exploratory Data Analysis (EDA)

## 🔹 What is EDA?
Exploratory Data Analysis (EDA) is the process of analyzing datasets to:
- Understand structure (rows, columns, data types)
- Summarize statistics (mean, median, min, max, std)
- Detect missing values, duplicates, and outliers
- Identify distributions and correlations
- Prepare data for modeling

---

## 🔹 Why EDA is Used
- ✅ Gain insights before modeling
- ✅ Ensure data quality
- ✅ Spot anomalies
- ✅ Guide feature selection
- ✅ Support better decisions

---

## 🔹 Common EDA Techniques
- **Descriptive statistics** → `df.describe()`, `df.info()`
- **Data visualization** → histograms, scatter plots, box plots
- **Correlation analysis** → heatmaps, pairplots
- **Missing values check** → `df.isnull().sum()`
- **Value counts** → `df['column'].value_counts()`

---

## 🔹 Example in Python
```python
import pandas as pd

# Sample dataset
df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [25, 30, 35, 40],
    "Salary": [50000, 60000, 70000, 80000]
})

# Quick EDA steps
print(df.head())        # First 5 rows
print(df.info())        # Data types + nulls
print(df.describe())    # Summary statistics
print(df.isnull().sum()) # Missing values check
print(df['Age'].value_counts()) # Frequency of ages

````python
# 📊 Datasets for EDA Practice

## 🔹 Basic Level
- **Titanic Dataset**  
  Columns: Passenger info (Age, Sex, Class, Survived)  
  Use cases: Missing values, categorical encoding, survival analysis  

- **Iris Dataset**  
  Columns: Sepal length, Sepal width, Petal length, Petal width, Species  
  Use cases: Classification, visualization, correlation  

- **Students Performance Dataset**  
  Columns: Gender, Parental education, Test scores  
  Use cases: Descriptive stats, distributions, group comparisons  

---

## 🔹 Medium Level
- **Netflix Movies & TV Shows Dataset**  
  Columns: Title, Genre, Release year, Country, Duration  
  Use cases: Text cleaning, categorical analysis, trend analysis  

- **Airbnb Listings Dataset**  
  Columns: Location, Price, Room type, Reviews  
  Use cases: Outlier detection, price distribution, geospatial analysis  

- **Sales Data (Superstore)**  
  Columns: Order ID, Product, Category, Sales, Profit, Region  
  Use cases: Aggregations, pivot tables, profit analysis  

---

## 🔹 Hard Level
- **NYC Taxi Trips Dataset**  
  Columns: Pickup time, Drop-off time, Distance, Fare, Location  
  Use cases: Time-series analysis, geospatial mapping, anomaly detection  

- **Healthcare Dataset (Diabetes / Heart Disease)**  
  Columns: Patient info, Medical test results, Outcome  
  Use cases: Feature engineering, correlation, predictive modeling  

- **Financial Stock Market Dataset**  
  Columns: Date, Open, Close, High, Low, Volume  
  Use cases: Trend analysis, volatility, moving averages, forecasting  

---

## 🔹 Example in Python
```python
import pandas as pd

# Load Titanic dataset (basic level)
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

# Quick EDA
print(df.head())        # First 5 rows
print(df.info())        # Data types + nulls
print(df.describe())    # Summary statistics
print(df['Survived'].value_counts()) # Survival distribution
````
