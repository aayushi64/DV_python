```python
# Parameters vs Arguments in Python
#Parameter → The variable name you put inside the function definition.
#Argument → The actual value you pass when calling the function.
# Function definition (num is a PARAMETER)
def check_even_odd(num):
    if num % 2 == 0:
        print(f"{num} is Even")
    else:
        print(f"{num} is Odd")

# Function call (7 is an ARGUMENT)
check_even_odd(7)
```

```python
#A positional argument is when you pass values to a function in the exact order that the parameters are defined.
#The position (order) matters.
#Python matches arguments to parameters based on their place in the function call.
# Function with two parameters
def add_numbers(a, b):
    return a + b

# Calling with positional arguments
result1 = add_numbers(5, 10)   # a = 5, b = 10
result2 = add_numbers(20, 30)  # a = 20, b = 30

print("Result 1:", result1)
print("Result 2:", result2)
```

```python
#A default parameter means you give a parameter a predefined value in the function definition.
# If the caller doesn’t provide an argument, Python uses the default.
# If the caller does provide an argument, it overrides the default.
def areacircle(r, pie= 3.14):
    return(r*pie*r)
areacircle(7)
#yaaha pr overwrite kr dega pie ki value ko 
areacircle(7,3.1436)
```

```python
# When you call a function, you can pass values by naming the parameters explicitly.
#Order doesn’t matter.
#Python matches values to parameters using the name.
def greet(name, age):
    print(f"Hello {name}, you are {age} years old.")

# Calling with keyword arguments
greet(name="Arisha", age=19)
greet(age=21, name="Rahul")   # order swapped, still works
```

```python
#Sometimes you don’t know how many values a user will pass to a function.
#You can use *args → for arbitrary positional arguments.
#You can use **kwargs → for arbitrary keyword arguments.
def add_numbers(*args):
    return sum(args)

print(add_numbers(1, 2, 3))        # Output: 6
print(add_numbers(10, 20, 30, 40)) # Output: 100
```

```python
#simple arbitrary is defined by using *args or *anything 
#args are stored in form in form of tupple

def addition(*args):
    print(type(args))   # args is a tuple
    total = 0
    for i in args:
        total += i
    return total

# Call the function with arguments
result = addition(10, 20, 30)
print("Sum =", result)

```

```python
#keyword arbitary 
#In Python, you use **kwargs to accept any number of keyword arguments.
#These arguments are packed into a dictionary (key-value pairs).
#Useful when you don’t know beforehand which named arguments will be passed.
def student_info(**kwargs):
    print(type(kwargs))   # kwargs is a dict
    for key, value in kwargs.items():
        print(f"{key}: {value}")

# Calling with keyword arguments
student_info(name="Arisha", age=19, course="B.Tech")

```

```python
def student_info(**kwargs):
    print(type(kwargs))   # kwargs is a dict
    for i in kwargs.keys():
        print(i)

# Calling with keyword arguments
student_info(name="Arisha", age=19, course="B.Tech")
```

```python
# method vs function

#Function
#A function is a block of code that performs a task.
#Defined using def keyword.
#Can exist independently (outside of any class).

#method 
#A method is also a function, but it is associated with an object/class.
#Called using dot notation on an object.
#It automatically takes the object (self) as its first parameter.


#Function → Independent, can be called directly.

#Method → Belongs to a class/object, called with dot (.).
```

```python
def is_perfect(n):
    s = 0
    for i in range(1, n):
        if n % i == 0:
            s = s + i
    if s == n:
        return True
    else:
        return False

for num in range(1, 1001):
    if is_perfect(num):
        print(num)
```

```python
# to convert the temp into celcius or fahrenheit using user provide and covert to respective using functions 
def to_celsius(f):
    return (f - 32) * 5/9

def to_fahrenheit(c):
    return (c * 9/5) + 32

choice = input("Convert to (C/F): ")

if choice.upper() == "C":
    f = float(input("Enter Fahrenheit: "))
    print("Celsius =", to_celsius(f))
elif choice.upper() == "F":
    c = float(input("Enter Celsius: "))
    print("Fahrenheit =", to_fahrenheit(c))
else:
    print("Invalid choice")
```

```python
def is_strong_password(password):
    if len(password) < 8:
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char.islower() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any(char in '!@#$%^&()_' for char in password):
        return False
    return True


password = input("Enter your password: ")  # keep it as string
if is_strong_password(password):
    print("Password is strong ✅")
else:
    print("Password is weak ❌")
```

```python
#cost calculation
def calculate_total_cost(cart):
    total_cost = 0
    for item in cart:
        total_cost += item['price'] * item['quantity']
    return total_cost


cart = [
    {'name': 'apple', 'price': 0.5, 'quantity': 4},
    {'name': 'banana', 'price': 0.6, 'quantity': 5},
    {'name': 'Orange', 'price': 0.7, 'quantity': 3}
]

total_cost = calculate_total_cost(cart)
print(f"Total Cost: ${total_cost:.2f}")  # Output: Total Cost: $7.10
```
