```python
# prime no
for num in range(1, 101):
    if num > 1:
        for i in range(2, num):
            if (num % i) == 0:
                break
        else:
            print(num)
```

```python
#sum of n no
n = 10
sum = 0
count = 0
for i in range(1, n + 1):
    sum += i
    count += 1
print("Sum of first", n, "natural numbers is:", sum)
print("Count of natural numbers is:", count)
```
