```python
#Defines the separator between multiple values (default is space " ").
a = 10
b = 15
c = 20
print(a, b, c, sep = "@")
# jese ki yaha print statement me humne sep = "@" define kiya hai toh 
# output me values ke beech me @ aa jayega.
#or by default end next line pe chala jata hai. bhale hi define krao ya na kro.
print("A", "B", "C", sep="-")
```

```python
#Defines what is printed at the end (default is newline \n).
#mtlb agr tum end ko define nhi kroge toh voh automatically new line pe chala jayega.
a = 10
b = 15
c = 20
print(a, b, c, sep = " @ ", end = " % ")
```
