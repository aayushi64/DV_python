```python
def isPalindrome(s):
    s = s.lower().replace(" ", "")
    return s == s[::-1]

print(isPalindrome("A man a plan a canal Panama"))
```

```python
def factorial(n):
    fact = 1
    for i in range(1, n + 1):
        fact = fact * i
    return fact

num = int(input("Enter a number: "))
print("Factorial =", factorial(num))
```

```python
# wap to make it recursion 
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

for i in range(10):
    print(fibonacci(i), end=" ")

```

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
```

```python
# sum of n no 
def sum(n):
    if n == 0:
        return 0
    return n + sum(n - 1)

print(sum(10))   # 55

```

```python
# using function only
def add(x, y, z):
    return x + y + z

print(add(12, 12, 12))   # Output: 36


# using lambda function
x = lambda a, b, c: a + b + c
print(x(2, 5, 6))        # Output: 13
```

```python
# map() applies a function to all items in a list

numbers = [1, 2, 3, 4, 5, 6]

def square(number):
    return number ** 2

print(square(6))   # Example direct call

# Using map with lambda
result = list(map(lambda x: x**2, numbers))
print(result)      # Output: [1, 4, 9, 16, 25, 36]
```
