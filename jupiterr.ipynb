```python
#short circuit operator
# and → If the first condition is False, Python won’t check the second one 
# (because the whole result must be False) 
#or → If the first condition is True, Python won’t check the second one 
# (because the whole result must be True).
# AND short-circuit
x = 0
if x != 0 and (10 / x > 1):   # second part won't run because x != 0 is False
    print("Valid")
else:
    print("Skipped safely")   # avoids ZeroDivisionError

# OR short-circuit
name = ""
if name or "Default":   # since name is empty, "Default" is used
    print("Result:", name or "Default")
```

```python
#Identity operator
#gives ans in true or false in "is and isnot"
#identity operators are used to check whether two variables point to the same object in memory, not just whether their values are equal.
#is → Returns True if both variables refer to the same object.
#is not → Returns True if they refer to different objects.
# Example 1: Same object
a = [1, 2, 3] #list hai ye!!
b = a
#inki bs refrence dekega
print(a is b)       # True (both point to the same list) # same refrence ko bolre
#inki bs value check krega
print(a == b)       # True (values are equal too) inki value bhi same hai

# Example 2: Different objects, same value
x = [1, 2, 3]
y = [1, 2, 3]
print(x is y)       # False (different objects in memory)
print(x == y)       # True (values are equal)

# Example 3: Using None
z = None
print(z is None)    # True (common Python idiom)
```

```python
#the id() function is used to get the unique identity number of an object. 
#This number represents the memory address where the object is stored 
# (though the exact value is implementation-dependent).
#if x = 10, y = 10 and z =10 toh python by default same memory mai ye utilise krega jise inka id same niklegi 
# disadvantage = For immutable objects (like integers, strings, tuples),
#  Python often reuses memory, so two variables with the same value may share the same id.
# Example 1: Same object
a = [1, 2, 3]
b = a
print(id(a))   # e.g., 140711234567888
print(id(b))   # same as a, because both point to the same list

# Example 2: Different objects with same value
x = [1, 2, 3]
y = [1, 2, 3]
print(id(x))   # different number
print(id(y))   # different number, even though values are equal

# Example 3: Immutable objects
p = 10
q = 10
print(id(p))   # same id (Python reuses small integers)
print(id(q))   # same id
```

```python
#shallow copy and deep copy
# shallow copy and deep copy are ways to duplicate objects, but they differ in how they handle nested (inner) objects.
# shallow : Creates a new object, but references of nested objects are shared.
# Changes in nested objects affect both copies.
import copy

list1 = [[1, 2], [3, 4]]
shallow = copy.copy(list1)

shallow[0][0] = 99
print(list1)     # [[99, 2], [3, 4]] → original changed
print(shallow)   # [[99, 2], [3, 4]]


 #deep copy : Creates a new object and recursively copies all nested objects.
 # Changes in nested objects do not affect the original.


list1 = [[1, 2], [3, 4]]
deep = copy.deepcopy(list1)

deep[0][0] = 99
print(list1)     # [[1, 2], [3, 4]] → original unchanged
print(deep)      # [[99, 2], [3, 4]]
```

```python
#the not in operator is used to check whether a value is absent from a sequence 
# (like a list, tuple, string, or set).
#Returns True if the element is not present.
# Returns False if the element is present.
# Example 1: List
numbers = [1, 2, 3, 4]
print(5 not in numbers)   # True (5 is not in the list)
print(2 not in numbers)   # False (2 is in the list)

# Example 2: String
text = "hello"
print("z" not in text)    # True (z not in "hello")
print("h" not in text)    # False (h is in "hello")

# Example 3: Tuple
tup = (10, 20, 30)
print(40 not in tup)      # True
print(20 not in tup)      # False
```
