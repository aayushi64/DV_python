# Topic 1 - plot 

🎯 Hierarchy Breakdown
Figure

Sabse top‑level container.

Ek figure ke andar multiple plots (axes) ho sakte hain.

Example: fig = plt.figure()

Axes

Ek individual plot area (x‑axis + y‑axis + data).

Har figure mein ek ya zyada axes ho sakte hain.

Example: ax = fig.add_subplot(1,1,1)

Axis

Axes ke andar horizontal (x) aur vertical (y) axis objects hote hain.

Ye ticks, labels, limits handle karte hain.

Example: ax.set_xlim(0,10)

Artist

Har visible element (lines, text, legends, titles, etc.) ek Artist object hota hai.

Matlab jo bhi tum dekhte ho plot mein, wo ek artist hai.

Example: ax.plot(x,y) → line artist create hota hai.

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5,6]
y = [2,4,6,8,10,12]

plt.figure(figsize=(5,3))   # Figure object (top-level container)
plt.plot(x,y)               # Line artist inside Axes
plt.title("Line Plot")      # Title artist
plt.show()                  # Render the Figure
```

```python
fig = plt.figure(figsize=(5,3))   # Figure object (top-level container)
ax.plot(x,y)               # Line artist inside Axes
ax.set_title("Line Plot")      # Title artist
ax.set_xlabel("X-axis")      # X-axis label
ax.set_ylabel("Y-axis")      # Y-axis label
plt.show()                  # Render the Figure
```

🔎 Difference:
Pyplot style → uses global functions (plt.subplot, plt.plot, plt.title). Quick and simple, but less control.

Object-Oriented style → works with explicit objects (fig, axes). More flexible, especially for complex layouts, customization, or multiple figures.

👉 Exam tip: Pyplot = shortcut, OO = professional/cleaner.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)

# Pyplot style
plt.figure(figsize=(6,4))
plt.subplot(1,2,1)   # first subplot
plt.plot(x, y, 'r')
plt.title("Sine - Pyplot")

plt.subplot(1,2,2)   # second subplot
plt.plot(x, np.cos(x), 'b')
plt.title("Cosine - Pyplot")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)

# Object-Oriented style
fig, axes = plt.subplots(1,2, figsize=(6,4))

axes[0].plot(x, y, 'r')
axes[0].set_title("Sine - OO")

axes[1].plot(x, np.cos(x), 'b')
axes[1].set_title("Cosine - OO")

fig.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Pyplot style (implicit canvas)
plt.figure(figsize=(5,3))   # creates canvas
plt.plot([1,2,3],[2,4,6], marker='o')
plt.title("Pyplot Canvas")
plt.show()


# Object-Oriented style (explicit canvas)
fig, ax = plt.subplots(figsize=(5,3))  # fig = canvas, ax = subplot
ax.plot([1,2,3],[2,4,6], marker='o')
ax.set_title("OO Canvas")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(1,7)
y = x**2

fig, ax = plt.subplots(figsize=(5,3))   # Correct way to create Figure + Axes
ax.plot(x, y, marker='o', color='#FF5733')  # Hex color code
ax.set(title="Quadratic Plot", xlabel="X-axis", ylabel="Y-axis")
# ax.set use kr ke hum title xlabel and ylabel ko set kr skte hai
# figsize shoule be in tuple format (width, height)
# and should be selected carefully 
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

days = np.arange(1, 8)
attendance = [90, 85, 88, 92, 95, 80, 87]

fig, ax = plt.subplots(figsize=(6,3.5))   # Correct way
ax.plot(days, attendance, marker='o')

ax.set(
    xlim=(1, 7),
    ylim=(75, 100),
    title="Weekly Attendance",
    xlabel="Days",
    ylabel="Attendance (%)"
)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

semesters = np.arange(1,7)
cgpa = [7.1 , 9.4, 5.6, 8.2, 7.8, 9.0]

fig, ax = plt.subplots(figsize=(6,3.5))   # Correct way
ax.plot(semesters, cgpa, marker='o', color='green')

ax.set_xticks(semesters)                         # X-axis ticks
ax.set_yticks(np.arange(6.5, 9.6, 0.5))          # Y-axis ticks
ax.grid(axis='y', linestyle='--', alpha=0.7)     # Grid lines

ax.set(
    xlim=(1, 6),
    ylim=(5, 10),
    title="Semester CGPA",
    xlabel="Semesters",
    ylabel="CGPA"
)

plt.show()
```
