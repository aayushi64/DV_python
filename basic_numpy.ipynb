# Array Dimensions in NumPy

## What is Dimension?
In NumPy, the **dimension** of an array refers to the number of axes (directions) it has.  
It is also called the **rank** of the array.

- **1D Array** → Single row/line of elements  
- **2D Array** → Table with rows and columns  
- **3D Array** → Cube-like structure with depth  

````python
import numpy as np

# 1D Array
arr1D = np.array([1, 2, 3, 4, 5])
print("1D Array:\n", arr1D)

# 2D Array
arr2D = np.array([[1, 2, 3], [4, 5, 6]])
print("\n2D Array:\n", arr2D)

# 3D Array
arr3D = np.array([[[1, 2], [3, 4]],[[5, 6], [7, 8]]])
print("\n3D Array:\n", arr3D)
````

````python
import numpy as np

# 1D Zero Array
arr1D = np.zeros(5)
print("1D Zero Array:\n", arr1D)

# 2D Zero Array
arr2D = np.zeros((2, 3))
print("\n2D Zero Array:\n", arr2D)

# 3D Zero Array
arr3D = np.zeros((2, 2, 2))
print("\n3D Zero Array:\n", arr3D)
````

# NumPy `arange()`

## What is `np.arange()`?
`np.arange()` is used to create arrays with evenly spaced values within a given range.  
It’s similar to Python’s built‑in `range()`, but it returns a NumPy array instead of a list.

## Syntax

- **start** → starting value (default = 0)  
- **stop** → end value (exclusive)  
- **step** → difference between values (default = 1)  
- **dtype** → optional, to set type (int/float)

## Examples
```python
import numpy as np



````python
# np.arange() 
# Simple range
arr1 = np.arange(5)
print(arr1)   # [0 1 2 3 4]

# Custom start and stop
arr2 = np.arange(2, 10)
print(arr2)   # [2 3 4 5 6 7 8 9]

# With step
arr3 = np.arange(1, 10, 2)
print(arr3)   # [1 3 5 7 9]

# Float step*
arr4 = np.arange(0, 1, 0.2)
print(arr4)   # [0.  0.2  0.4  0.6  0.8]
````

# NumPy `linspace()`

## What is `np.linspace()`?
`np.linspace()` is used to create arrays with evenly spaced values **between two numbers**.  
Unlike `np.arange()`, which uses a step size, `np.linspace()` specifies the **number of values** you want.

## Syntax

- **start** → starting value  
- **stop** → ending value  
- **num** → number of values to generate (default = 50)  
- **endpoint** → if True, includes the stop value (default = True)  
- **dtype** → optional, to set type (int/float)

# DIFFERENCE BTW ARANGE AND LINESPACE
👉 Quick difference:  
- `np.arange(start, stop, step)` → control **step size**  
- `np.linspace(start, stop, num)` → control **number of values**  

````python
#np.linspace()
# 5 numbers between 0 and 10
arr = np.linspace(0, 10, dtype=int)
print(arr)
````

````python
arr = np.linspace(0, 10, 11)
print(arr)  
````

The reason you get 50 values by default is because NumPy’s designers 
set num=50 as the default parameter in np.linspace().

# ⚡ Vectorization in NumPy

## 📌 Definition
Vectorization means performing operations on **entire arrays** instead of looping through elements one by one.  
NumPy uses optimized C code internally, making these operations **faster, cleaner, and more memory‑efficient**.

---

## 🔎 Without Vectorization (Python Lists)
```python
list1 = [1, 2, 3, 4, 5]
list2 = [10, 20, 30, 40, 50]

# Manual loop addition
result_list = [x + y for x, y in zip(list1, list2)]
print("List result:", result_list)

````python
import numpy as np

arr1 = np.array([1, 2, 3, 4, 5])
arr2 = np.array([10, 20, 30, 40, 50])

# Vectorized addition
result_array = arr1 + arr2
print("Array result:", result_array)
````

````python
import numpy as np

# Arrays
arr1 = np.array([1, 2, 3, 4, 5])
arr2 = np.array([10, 20, 30, 40, 50])

# Addition
print("Addition:", arr1 + arr2)

# Subtraction
print("Subtraction:", arr2 - arr1)

# Multiplication
print("Multiplication:", arr1 * arr2)

# Division
print("Division:", arr2 / arr1)

# Power
print("Power:", arr1 ** 2)

# Square Root
print("Square Root:", np.sqrt(arr2))
````

With lists, you need explicit loops (for or zip) → slower.

With NumPy arrays, you get true vectorization (arr1 + arr2) → faster and cleaner.

````python
# without vectorization (using Python lists)
list1 = [1, 2, 3, 4, 5]
list2 = [10, 20, 30, 40, 50]
# Using list comprehension for addition
result_list = [x + y for x, y in zip(list1, list2)]
print("List result:", result_list)
````

````python
# Python Lists (element-wise addition using for loop)
list1 = [1, 2, 3, 4, 5]
list2 = [10, 20, 30, 40, 50]

# Using for loop for addition
result_list = []
for i in range(len(list1)):
    result_list.append(list1[i] + list2[i])

print("List result:", result_list)
````

````python
import numpy as np
import time

# Python Lists
lst = [x for x in range(100000)]
start = time.time()
lst2 = [x + 1 for x in lst]   # manual loop
end = time.time()
print("Time taken using Python lists:", (end - start) * 1000, "ms")

# NumPy Arrays
arr = np.arange(100000)       # proper array from 0 → 99999
start1 = time.time()
arr2 = arr + 1                # vectorized addition
end1 = time.time()
print("Time taken using NumPy arrays:", (end1 - start1) * 1000, "ms")
````

# 📡 NumPy Broadcasting

## 📌 Definition
Broadcasting allows NumPy to perform operations on arrays of **different shapes** by automatically expanding the smaller array to match the larger one.

---

# 📡 Broadcasting in NumPy

## 📌 Why Broadcasting?
- To perform operations between arrays of **different shapes** without writing loops.
- Saves time and makes code shorter and cleaner.
- Lets you mix scalars, vectors, and matrices easily.

---

## 🛠 What Broadcasting Does
- **Expands smaller arrays** to match the shape of larger ones.
- Applies the operation element‑wise once shapes are compatible.

---


## 🛠 Example 1: Scalar + Array
```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])
result = arr + 10   # scalar broadcasted
print(result)

````python
import numpy as np

# Example 1: Scalar + Array
arr = np.array([1, 2, 3, 4, 5])
print("Scalar + Array:", arr + 10)

# Example 2: 2D + 1D Array
A = np.array([[1, 2, 3],
              [4, 5, 6]])
B = np.array([10, 20, 30])
print("2D + 1D Array:\n", A + B)

# Example 3: 3D + 1D Array
X = np.ones((2, 3, 4))   # shape (2,3,4)
Y = np.array([1, 2, 3, 4])  # shape (4,)
print("3D + 1D Array:\n", X * Y)

# Example 4: Column Vector + Row Vector
row = np.array([1, 2, 3])          # shape (3,)
col = np.array([[10], [20], [30]]) # shape (3,1)
print("Column + Row:\n", col + row)
````

# 🎲 NumPy Random Number Generation Cheat Sheet

## 1. Basic Random Functions
- `np.random.rand(d0, d1, ..., dn)`  
  → Random floats in [0, 1), uniform distribution.  
  Example: `np.random.rand(2,3)` → 2x3 matrix of random floats.

- `np.random.randn(d0, d1, ..., dn)`  
  → Random samples from standard normal distribution (mean=0, variance=1).  
  Example: `np.random.randn(4)` → 1D array of 4 values.

- `np.random.randint(low, high=None, size=None)`  
  → Random integers from `low` (inclusive) to `high` (exclusive).  
  Example: `np.random.randint(1,10,5)` → 5 random integers between 1–9.

## 2. Choice & Permutation
- `np.random.choice(a, size=None, replace=True, p=None)`  
  → Random sample from array or range.  
  Example: `np.random.choice([1,2,3,4], 2)` → Pick 2 random elements.

- `np.random.permutation(x)`  
  → Randomly permute sequence or array.  
  Example: `np.random.permutation([1,2,3,4])` → Shuffled array.

- `np.random.shuffle(x)`  
  → Shuffle array in place.  
  Example: `arr = np.array([1,2,3,4]); np.random.shuffle(arr)`.

## 3. Distributions
- `np.random.uniform(low=0.0, high=1.0, size=None)`  
  → Uniform distribution between `low` and `high`.  
  Example: `np.random.uniform(5,10,3)` → 3 floats between 5–10.

- `np.random.normal(loc=0.0, scale=1.0, size=None)`  
  → Normal (Gaussian) distribution with mean=`loc`, std=`scale`.  
  Example: `np.random.normal(50, 10, 5)` → 5 values around mean 50.

- `np.random.binomial(n, p, size=None)`  
  → Binomial distribution (number of successes in `n` trials).  
  Example: `np.random.binomial(10, 0.5, 5)` → 5 samples.

- `np.random.poisson(lam=1.0, size=None)`  
  → Poisson distribution with expected value `lam`.  
  Example: `np.random.poisson(3, 5)` → 5 samples.

- `np.random.exponential(scale=1.0, size=None)`  
  → Exponential distribution.  
  Example: `np.random.exponential(2, 5)` → 5 samples.

## 4. Random State Control
- `np.random.seed(seed)`  
  → Set seed for reproducibility.  
  Example: `np.random.seed(42)` → Same random results every run.

- `np.random.default_rng(seed)`  
  → New recommended random generator (NumPy ≥1.17).  
  Example:

  rng = np.random.default_rng(42)
  rng.integers(1,10,5)
