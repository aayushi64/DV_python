# Pandas 🐼

## Definition
Pandas is an **open-source Python library** built for **data analysis and manipulation**.  
It provides fast, flexible, and expressive data structures like **Series** (1D) and **DataFrame** (2D, tabular) to work with structured data easily.

## What Pandas is Used For
- 📊 **Data Cleaning** → handle missing values, duplicates, formatting
- 📑 **Data Manipulation** → filtering, grouping, merging, reshaping
- 📈 **Data Analysis** → statistical summaries, aggregations
- 📂 **Data Import/Export** → read/write CSV, Excel, SQL, JSON
- 🔍 **Indexing & Selection** → powerful row/column access with `loc` and `iloc`
- ⚡ **Integration** → works seamlessly with NumPy, Matplotlib, and machine learning libraries

## Quick Example
```python
import pandas as pd

data = {'Name': ['A', 'B', 'C'], 'Age': [20, 21, 19]}
df = pd.DataFrame(data)

print(df)

📌 Pandas Series in Python
🔹 Definition
A Series is a one‑dimensional labeled array in Pandas that can hold any data type (integers, strings, floats, etc.).

````python
import pandas as pd

s = pd.Series([145, 142, 24, 13], name="counts")
print(s)
s.index
````

# NaN in Python (NumPy)

## Definition
- **NaN** stands for **Not a Number**.
- It represents missing or undefined numerical data.
- will ignore this value and execute as usuall
- pands deal with nan value directly while numpy cant

## Example in Python
```python
import numpy as np

# Create an array with NaN
arr = np.array([1, 2, np.nan, 4])

# Check for NaN values
print(np.isnan(arr))   # [False False  True False]

# Replace NaN with a number
arr = np.nan_to_num(arr, nan=0)
print(arr)             # [1. 2. 0. 4.]


# 📌 Load Data from CSV into Series (Python Pandas)

## 🔹 Definition
You can directly read a column from a CSV file into a **Series** using `pandas.read_csv()`.

---

## 🔹 Syntax
```python

# Load CSV file

# Convert a specific column into Series

print(s)
