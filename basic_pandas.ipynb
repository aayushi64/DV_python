# Pandas 🐼

## Definition
Pandas is an **open-source Python library** built for **data analysis and manipulation**.  
It provides fast, flexible, and expressive data structures like **Series** (1D) and **DataFrame** (2D, tabular) to work with structured data easily.

## What Pandas is Used For
- 📊 **Data Cleaning** → handle missing values, duplicates, formatting
- 📑 **Data Manipulation** → filtering, grouping, merging, reshaping
- 📈 **Data Analysis** → statistical summaries, aggregations
- 📂 **Data Import/Export** → read/write CSV, Excel, SQL, JSON
- 🔍 **Indexing & Selection** → powerful row/column access with `loc` and `iloc`
- ⚡ **Integration** → works seamlessly with NumPy, Matplotlib, and machine learning libraries

## Quick Example

# NaN in Python (NumPy)

## Definition
- **NaN** stands for **Not a Number**.
- It represents missing or undefined numerical data.
- will ignore this value and execute as usuall
- pands deal with nan value directly while numpy cant

## Example in Python
```python
import numpy as np

# Create an array with NaN
arr = np.array([1, 2, np.nan, 4])

# Check for NaN values
print(np.isnan(arr))   # [False False  True False]

# Replace NaN with a number
arr = np.nan_to_num(arr, nan=0)
print(arr)             # [1. 2. 0. 4.]

# Pandas Data Structures 📚

## 1. Series (1D)
- One‑dimensional labeled array
- Similar to a column in Excel
- Index + Values
- **Syntax:** `pd.Series(data, index)`

---

## 2. DataFrame (2D)
- Two‑dimensional labeled data structure
- Rows + Columns (like database table)
- Can hold heterogeneous data
- **Syntax:** `pd.DataFrame(data)`

---

## 3. Panel (3D) ⚠️ (Deprecated)
- Three‑dimensional data structure
- Items × Major_axis × Minor_axis
- Used for 3D data (like multiple DataFrames stacked)
- **Note:** Deprecated since Pandas 0.25 → use **MultiIndex DataFrame** or **xarray** instead.

---

## 4. Comparison Table 🔍

| Structure   | Dimension | Example Use Case                |
|-------------|-----------|---------------------------------|
| Series      | 1D        | Marks of one subject            |
| DataFrame   | 2D        | Student database (rows+columns) |
| Panel       | 3D        | Multiple DataFrames together    |

---

## 5. Analogy 💡
- **Series** → Single column  
- **DataFrame** → Full table  
- **Panel** → Collection of tables (like 3D cube)

---

## 6. Practical Notes
- **Series** → Best for 1D labeled data  
- **DataFrame** → Most commonly used (CSV, Excel, SQL)  
- **Panel** → Avoid, use MultiIndex instead

# Difference in NumPy vs Pandas (dtype handling)

## 1. NumPy Arrays
- Homogeneous data → sab elements same type hone chahiye
- Agar mix karoge (int + float + string), NumPy sabko ek hi common dtype mein convert karega
- Example:
```python
import numpy as np
arr = np.array([10, 20, 30])       # int64
arr2 = np.array([10, 20.5, 30])    # float64
arr3 = np.array([10, "A", 30])     # <U21 (string)

2. Pandas Series
Heterogeneous data allowed

Pandas dtype rules (Rule of Thumb):

All int → int64

All float → float64

All string → object

Mix with None → float64 (NaN handling)

Example:
import pandas as pd
s1 = pd.Series([10, 20, 30])          # int64
s2 = pd.Series([10, 20.5, 30])        # float64
s3 = pd.Series([10, "A", 30])         # object
s4 = pd.Series([10, None, 30])        # float64 (NaN)
import pandas as pd
s1 = pd.Series([10, 20, 30])          # int64
s2 = pd.Series([10, 20.5, 30])        # float64
s3 = pd.Series([10, "A", 30])         # object

3. Key Differences 🔍
Feature	NumPy Array	Pandas Series
Data type	Homogeneous (single dtype)	Flexible (can be object dtype)
None / NaN handling	None → object dtype	None → NaN (float64)
Labels (Index)	No labels, only positions	Has index labels
Use Case	Numerical computation	Tabular / labeled dat

4. Analogy 💡
NumPy → Pure math arrays (fast, strict types)

Pandas → Database‑like structures (flexible, labeled, handles missing values)




# NaN in NumPy vs Pandas

## 1. NumPy
- `np.nan` ek special floating‑point value hai (IEEE 754 standard).
- NumPy arrays mein agar `None` daaloge → dtype `object` ban jaata hai.
- Agar `np.nan` daaloge → dtype `float64` ban jaata hai.
- Example:
python
import numpy as np

arr1 = np.array([10, np.nan, 30])   # float64
arr2 = np.array([10, None, 30])     # object

## 2. Pandas
- Pandas None ko automatically NaN treat karta hai jab numeric Series ho.
- Isliye None → NaN → dtype float64.
- Pandas ke paas special functions hain NaN handle karne ke liye:
- pd.isna() / pd.notna()
- Series.fillna(value)
- Series.dropna()
- Series.mean() (ignore NaN automatically)
Example
python
import pandas as pd

s = pd.Series([10, None, 30])
print(s)
print("dtype:", s.dtype)
print("Check NaN:", pd.isna(s))
print("Fill NaN:", s.fillna(0))

## 3. Key Differences 🔍
# Key Differences: NaN Handling in NumPy vs Pandas

| Feature              | NumPy                        | Pandas                        |
|----------------------|------------------------------|-------------------------------|
| None handling        | Becomes `object` dtype       | Converts to `NaN` (float64)   |
| NaN handling         | Stays `float64`, limited funcs | Rich functions (`fillna`, `dropna`, `mean`, etc.) |
| Missing data support | Basic                        | Advanced, database‑like        |


## 4. Analogy 💡
- >NumPy → NaN is just a float placeholder.
- >Pandas → NaN is a full missing‑data marker with powerful handling tools.
