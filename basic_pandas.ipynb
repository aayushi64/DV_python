# NaN in NumPy vs Pandas

## 1. NumPy
- `np.nan` ek special floating‑point value hai (IEEE 754 standard).
- NumPy arrays mein agar `None` daaloge → dtype `object` ban jaata hai.
- Agar `np.nan` daaloge → dtype `float64` ban jaata hai.
- Example:
python
import numpy as np

arr1 = np.array([10, np.nan, 30])   # float64
arr2 = np.array([10, None, 30])     # object

## 2. Pandas
- Pandas None ko automatically NaN treat karta hai jab numeric Series ho.
- Isliye None → NaN → dtype float64.
- Pandas ke paas special functions hain NaN handle karne ke liye:
- pd.isna() / pd.notna()
- Series.fillna(value)
- Series.dropna()
- Series.mean() (ignore NaN automatically)
Example
python
import pandas as pd

s = pd.Series([10, None, 30])
print(s)
print("dtype:", s.dtype)
print("Check NaN:", pd.isna(s))
print("Fill NaN:", s.fillna(0))

## 3. Key Differences 🔍
# Key Differences: NaN Handling in NumPy vs Pandas

| Feature              | NumPy                        | Pandas                        |
|----------------------|------------------------------|-------------------------------|
| None handling        | Becomes `object` dtype       | Converts to `NaN` (float64)   |
| NaN handling         | Stays `float64`, limited funcs | Rich functions (`fillna`, `dropna`, `mean`, etc.) |
| Missing data support | Basic                        | Advanced, database‑like        |


## 4. Analogy 💡
- >NumPy → NaN is just a float placeholder.
- >Pandas → NaN is a full missing‑data marker with powerful handling tools.

# 📌 Pandas Indexing Methods

## 🔹 `.loc` → Label-based indexing
- Access by **row/column labels**.
- Works with names in the index.

## 🔹 `.iloc` → Position-based indexing
- Access by **integer positions** (0, 1, 2, …).
- Similar to array-style indexing.

## 🔹 `.at` → Fast label-based scalar access
- Optimized for **single value** access by label.
- Faster than `.loc` when retrieving/updating one element.

## 🔹 `.iat` → Fast position-based scalar access
- Optimized for **single value** access by position.
- Faster than `.iloc` when retrieving/updating one element.

---

## ⚡ Comparison Table

| Method  | Basis            | Multiple values | Single value optimized | Example        |
|---------|------------------|-----------------|------------------------|----------------|
| `.loc`  | Label-based      | ✅              | ❌                     | `s.loc[2003]`  |
| `.iloc` | Position-based   | ✅              | ❌                     | `s.iloc[2]`    |
| `.at`   | Label-based      | ❌              | ✅                     | `s.at[2002]`   |
| `.iat`  | Position-based   | ❌              | ✅                     | `s.iat[0]`     |

---

# 🔄 Reset Index in Pandas

## Definition
- `reset_index()` is used to **reset the index of a DataFrame or Series** back to the default integer index (0, 1, 2, …).
- It is helpful when:
  - You have a custom index (like labels or after filtering).
  - You want to convert the index into a column.
  - You need a clean sequential index again.

---

# 🔄 Pandas reset_index()

## When to Use
- Useful when the **index is treated as a column** (e.g., after filtering or grouping).
- Needed when the **index is meaningless** and should be reset to the default integer index before another operation.

---

# 🔄 Pandas reset_index() Parameters

## drop
- **drop=True** → removes the old index instead of keeping it as a column.
- **drop=False** (default) → keeps the old index as a new column in the DataFrame.

## inplace
- **inplace=True** → modifies the DataFrame directly without creating a copy.
- **inplace=False** (default) → returns a new DataFrame with the reset index, leaving the original unchanged.

---
## Syntax
```python
df.reset_index(drop=False, inplace=False)


````python
import pandas as pd

# Create Series with custom index and index name
s = pd.Series(
    [1, 2, 3, 4],
    name="f1",
    index=pd.Index(["r1", "r2", "r3", "r4"], name="idx")
)

print(s)
print("dtype:", s.dtype)
print("Index:", s.index)
print("Index name:", s.index.name)
````

````python
import pandas as pd

# Create Series with custom index
s = pd.Series([10, 20, 30, 40], index=["p", "q", "r", "s"], name="values")

print("Original Series:\n", s)

# Reset index -> converts to DataFrame
df = s.reset_index()
print(type(s))
print(type(df))

print("\nDataFrame after reset_index:\n", df)
````

````python
import pandas as pd

# Create Series with custom index
s = pd.Series([10, 20, 30, 40], index=["p", "q", "r", "s"], name="f1")

print("Original Series:\n", s)

# Reset index with drop=True
df = s.reset_index(drop=True)
print(type(s))
print(type(df))

print("\nDataFrame after reset_index(drop=True):\n", df)
````

# 🔄 Pandas reset_index() — drop Parameter

## drop=False (default)
- Keeps the old index as a new column in the DataFrame.
- Useful when the index has meaning and you want to preserve it.
# 🔄 Pandas reset_index() — drop=True

## Behavior
- **Removes the old index completely.**
- Only the default integer index remains.
- Useful when the old index is meaningless and you want a clean DataFrame.



### Example
```python
import pandas as pd

s = pd.Series([10, 20, 30, 40], index=["p", "q", "r", "s"], name="f1")
df = s.reset_index(drop=False)
print(df)

import pandas as pd

s = pd.Series([10, 20, 30, 40], index=["p", "q", "r", "s"], name="f1")
df = s.reset_index(drop=True)
print(df)


````python
import pandas as pd

# Create Series
s = pd.Series([10, 20, 30, 40], index=["p", "q", "r", "s"], name="f1")

print("Original Series:\n", s)

# reset_index with inplace=False (default)
df1 = s.reset_index()
print("\nAfter reset_index (inplace=False):\n", df1)
print("\nOriginal Series still same:\n", s)

# reset_index with inplace=True
s.reset_index(drop=True, inplace=True)
print("\nAfter reset_index (inplace=True):\n", s)
````
