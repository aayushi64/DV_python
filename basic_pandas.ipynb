# NaN in NumPy vs Pandas

## 1. NumPy
- `np.nan` ek special floating‑point value hai (IEEE 754 standard).
- NumPy arrays mein agar `None` daaloge → dtype `object` ban jaata hai.
- Agar `np.nan` daaloge → dtype `float64` ban jaata hai.
- Example:
python
import numpy as np

arr1 = np.array([10, np.nan, 30])   # float64
arr2 = np.array([10, None, 30])     # object

## 2. Pandas
- Pandas None ko automatically NaN treat karta hai jab numeric Series ho.
- Isliye None → NaN → dtype float64.
- Pandas ke paas special functions hain NaN handle karne ke liye:
- pd.isna() / pd.notna()
- Series.fillna(value)
- Series.dropna()
- Series.mean() (ignore NaN automatically)
Example
python
import pandas as pd

s = pd.Series([10, None, 30])
print(s)
print("dtype:", s.dtype)
print("Check NaN:", pd.isna(s))
print("Fill NaN:", s.fillna(0))

## 3. Key Differences 🔍
# Key Differences: NaN Handling in NumPy vs Pandas

| Feature              | NumPy                        | Pandas                        |
|----------------------|------------------------------|-------------------------------|
| None handling        | Becomes `object` dtype       | Converts to `NaN` (float64)   |
| NaN handling         | Stays `float64`, limited funcs | Rich functions (`fillna`, `dropna`, `mean`, etc.) |
| Missing data support | Basic                        | Advanced, database‑like        |


## 4. Analogy 💡
- >NumPy → NaN is just a float placeholder.
- >Pandas → NaN is a full missing‑data marker with powerful handling tools.

# 📌 Pandas Indexing Methods

## 🔹 `.loc` → Label-based indexing
- Access by **row/column labels**.
- Works with names in the index.

## 🔹 `.iloc` → Position-based indexing
- Access by **integer positions** (0, 1, 2, …).
- Similar to array-style indexing.

## 🔹 `.at` → Fast label-based scalar access
- Optimized for **single value** access by label.
- Faster than `.loc` when retrieving/updating one element.

## 🔹 `.iat` → Fast position-based scalar access
- Optimized for **single value** access by position.
- Faster than `.iloc` when retrieving/updating one element.

---

## ⚡ Comparison Table

| Method  | Basis            | Multiple values | Single value optimized | Example        |
|---------|------------------|-----------------|------------------------|----------------|
| `.loc`  | Label-based      | ✅              | ❌                     | `s.loc[2003]`  |
| `.iloc` | Position-based   | ✅              | ❌                     | `s.iloc[2]`    |
| `.at`   | Label-based      | ❌              | ✅                     | `s.at[2002]`   |
| `.iat`  | Position-based   | ❌              | ✅                     | `s.iat[0]`     |

---

# 🔄 Reset Index in Pandas

## Definition
- `reset_index()` is used to **reset the index of a DataFrame or Series** back to the default integer index (0, 1, 2, …).
- It is helpful when:
  - You have a custom index (like labels or after filtering).
  - You want to convert the index into a column.
  - You need a clean sequential index again.

---

# 🔄 Pandas reset_index()

## When to Use
- Useful when the **index is treated as a column** (e.g., after filtering or grouping).
- Needed when the **index is meaningless** and should be reset to the default integer index before another operation.

---

# 🔄 Pandas reset_index() Parameters

## drop
- **drop=True** → removes the old index instead of keeping it as a column.
- **drop=False** (default) → keeps the old index as a new column in the DataFrame.

## inplace
- **inplace=True** → modifies the DataFrame directly without creating a copy.
- **inplace=False** (default) → returns a new DataFrame with the reset index, leaving the original unchanged.

---
## Syntax
```python
df.reset_index(drop=False, inplace=False)


````python
import pandas as pd

# Create Series with custom index and index name
s = pd.Series(
    [1, 2, 3, 4],
    name="f1",
    index=pd.Index(["r1", "r2", "r3", "r4"], name="idx")
)

print(s)
print("dtype:", s.dtype)
print("Index:", s.index)
print("Index name:", s.index.name)
````

````python
import pandas as pd

# Create Series with custom index
s = pd.Series([10, 20, 30, 40], index=["p", "q", "r", "s"], name="values")

print("Original Series:\n", s)

# Reset index -> converts to DataFrame
df = s.reset_index()
print(type(s))
print(type(df))

print("\nDataFrame after reset_index:\n", df)
````

````python
import pandas as pd

# Create Series with custom index
s = pd.Series([10, 20, 30, 40], index=["p", "q", "r", "s"], name="f1")

print("Original Series:\n", s)

# Reset index with drop=True
df = s.reset_index(drop=True)
print(type(s))
print(type(df))

print("\nDataFrame after reset_index(drop=True):\n", df)
````

# 🔄 Pandas reset_index() — drop Parameter

## drop=False (default)
- Keeps the old index as a new column in the DataFrame.
- Useful when the index has meaning and you want to preserve it.
# 🔄 Pandas reset_index() — drop=True

## Behavior
- **Removes the old index completely.**
- Only the default integer index remains.
- Useful when the old index is meaningless and you want a clean DataFrame.



### Example
```python
import pandas as pd

s = pd.Series([10, 20, 30, 40], index=["p", "q", "r", "s"], name="f1")
df = s.reset_index(drop=False)
print(df)

import pandas as pd

s = pd.Series([10, 20, 30, 40], index=["p", "q", "r", "s"], name="f1")
df = s.reset_index(drop=True)
print(df)


````python
import pandas as pd

# Create Series
s = pd.Series([10, 20, 30, 40], index=["p", "q", "r", "s"], name="f1")

print("Original Series:\n", s)

# reset_index with inplace=False (default)
df1 = s.reset_index()
print("\nAfter reset_index (inplace=False):\n", df1)
print("\nOriginal Series still same:\n", s)

# reset_index with inplace=True
s.reset_index(drop=True, inplace=True)
print("\nAfter reset_index (inplace=True):\n", s)
````

# 🔹 Pandas `Series.unique()`

The `unique()` method in Pandas is used to return the **distinct values** from a Series, removing duplicates.

## 📋 Key Points
- Returns a **NumPy array** of unique values.  
- Order of appearance is preserved (first occurrence comes first).  
- Useful for quickly identifying categories or distinct entries in a dataset.

---

# 🔎 Comparison of Pandas Methods

| Method             | Purpose |
|--------------------|---------|
| `s.unique()`       | Returns distinct values only |
| `s.value_counts()` | Returns unique values with their frequency |
| `s.nunique()`      | Returns the count of unique values |

---  

## 🖥️ Example
```python
import pandas as pd

# 🔹 Pandas `drop_duplicates()`

The `drop_duplicates()` method in Pandas is used to remove **duplicate rows or values** from a DataFrame or Series.

## 📋 Key Points
- Removes repeated entries and keeps only the first occurrence by default.  
- Can be applied to specific columns using the `subset` parameter.  
- The `keep` parameter controls which duplicates to retain (`first`, `last`, or `False` to drop all).  
- The `inplace` parameter allows modification of the original object without creating a copy.  

## 🖥️ Example
```python
import pandas as pd

data = pd.DataFrame({
    'A': [1, 2, 2, 3, 4, 4],
    'B': ['x', 'y', 'y', 'z', 'x', 'x']
})

print(data.drop_duplicates())


````python
import pandas as pd
import numpy as np

# Sample Series with NaN values
s = pd.Series([1, 2, 2, np.nan, 3, 4, 4, np.nan, 5, 5, 5], name="MySeries")

# Drop duplicates (keeps first occurrence, including NaN)
print("Drop duplicates:")
print(s.drop_duplicates())

# Unique values (returns NumPy array, includes NaN)
print("\nUnique values:")
print(s.unique())

# Frequency of each value (NaN excluded by default, use dropna=False to include)
print("\nValue counts (excluding NaN):")
print(s.value_counts())

print("\nValue counts (including NaN):")
print(s.value_counts(dropna=False))

# Show dtype and name
print("\nSeries dtype:", s.dtype)
print("Series name:", s.name)

````

# 🔹 Pandas: Drop Duplicate Index Entries

In Pandas, you can remove duplicate index entries using the `index.duplicated()` method combined with boolean indexing.

## 📋 Key Points
- `df.index.duplicated()` returns a boolean array marking duplicates as `True`.  
- Using `~` (NOT operator) keeps only the first occurrence of each index value.  
- This ensures a DataFrame has unique index labels.  

## 🖥️ Example
```python
import pandas as pd

# Sample DataFrame with duplicate index entries
df = pd.DataFrame({
    "A": [10, 20, 30, 40],
    "B": ["x", "y", "y", "z"]
}, index=["p", "q", "q", "r"])

print("Original DataFrame:")
print(df)

# Drop duplicate index entries (keep first occurrence)
df_clean = df[~df.index.duplicated(keep='first')]

print("\nAfter dropping duplicate index entries:")
print(df_clean)


````python
import pandas as pd

# Sample Series with duplicate index entries
s = pd.Series([300, 350, 30, 46], index=['falcon', 'falcon', 'hawk', 'hawk'], name="BirdData")

print("Original Series:")
print(s)

# Drop duplicate index entries (keep first occurrence)
s_clean = s[~s.index.duplicated(keep='first')]

print("\nAfter dropping duplicate index entries:")
print(s_clean)

# Show dtype and name
print("\nSeries dtype:", s_clean.dtype)
print("Series name:", s_clean.name)
````

# Python Built-in Functions: map(), zip(), filter()

## Comparison Table

| Function  | Purpose                                | Returns         | Example Use Case                  |
|-----------|----------------------------------------|-----------------|-----------------------------------|
| `map()`   | Apply function to each element         | Iterator (map)  | Square all numbers in a list      |
| `zip()`   | Combine iterables element‑wise         | Iterator (zip)  | Pair names with marks             |
| `filter()`| Select elements based on condition     | Iterator (filter)| Keep only even numbers from list |

---

## Examples

```python
# map() → transform each element
nums = [1, 2, 3, 4]
squares = list(map(lambda x: x**2, nums))  # #square each number
print(squares)  # [1, 4, 9, 16]

# zip() → combine iterables element-wise
names = ["Arisha", "Pihu", "Aditi"]
marks = [85, 90, 78]
combined = list(zip(names, marks))  # #pair names with marks
print(combined)  # [('Arisha', 85), ('Pihu', 90), ('Aditi', 78)]

# filter() → select elements based on condition
nums = [10, 25, 30, 45, 50]
evens = list(filter(lambda x: x % 2 == 0, nums))  # #keep only even numbers
print(evens)  # [10, 30, 50]


````python
import pandas as pd

s = pd.Series([1, 2, 3, None])

# Apply function to every element
print(s.map(lambda x: x**2 if x is not None else x))

# Apply dict mapping
print(s.map({1: "one", 2: "two", 3: "three"}))

# Apply with ignore NaN
print(s.map(lambda x: f"val={x}", na_action="ignore"))
````

# Pandas Series.map() with Another Series

## Concept
- `map()` can take another **Series** as argument.
- Any value of the calling Series that matches the **index** of the mapping Series will be mapped.
- The **index of the calling Series remains same**, but its **values get updated** to the corresponding mapped values.
- Non‑matching values → `NaN`.

---

````python
import pandas as pd

# Series 1: product names
products = pd.Series(["laptop", "mobile", "tablet", "camera"], index=["p1","p2","p3","p4"])

# Series 2: mapping Series (index = product, values = price)
prices = pd.Series([200, 300, 150], index=["laptop", "mobile", "tablet"])

# Map products → prices
mapped = products.map(prices)
print(mapped)
````
